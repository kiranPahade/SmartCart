```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
a = pd.read_csv("/content/smartcart_customers.csv")
a
```

```python
a.isnull().sum()
```

```python
a['Income'] = a['Income'].fillna(a['Income'].median())
a.isnull().sum()
```

```python
a.columns
```

```python
a["Age"] = 2026-a["Year_Birth"]
```

```python
a.head()
```

```python
a["Dt_Customer"] = pd.to_datetime(a["Dt_Customer"],dayfirst=True)
```

```python
reference_date = a["Dt_Customer"].max()
```

```python
reference_date
```

```python
a["Total_Days"] = (reference_date-a["Dt_Customer"]).dt.days
```

```python
a.head()
```

```python
a.columns
```

```python
a["Total_Spent_Amt"]=a['MntFruits']+a['MntGoldProds']+a['MntMeatProducts']+a['MntSweetProducts']+a['MntWines']+a['MntFishProducts']
```

```python
a.head()
```

```python
a["Total_Childs"] = a['Kidhome']+a['Teenhome']
```

```python
a.head()
```

```python
a['Education'].value_counts()
```

```python
a['Education'] = a['Education'].replace({"Graduation":"Graduate","PhD":"Graduate","Master":"Graduate","2n Cycle":"undergraduate","Basic":"undergraduate"})
```

```python
a['Education'].value_counts()
```

```python
a['Marital_Status'].value_counts()
```

```python
a["Living"] = a['Marital_Status'].replace({"Married":"P","Together":"P","Single":"A","Divorced":"A","Widow":"A","Alone":"A","Absurd":"A","YOLO":"A"})
```

```python
a['Living'].value_counts()
```

```python
a.columns
```

```python
V= ['ID',"Year_Birth","Marital_Status","Kidhome","Teenhome","Dt_Customer"]
V2 = ['MntWines', 'MntFruits',
       'MntMeatProducts', 'MntFishProducts', 'MntSweetProducts',
       'MntGoldProds']
```

```python
Both_Column = V + V2
```

```python
Both_Column
```

```python
a_Cleaned = a.drop(columns=Both_Column,axis=1)
```

```python
a_Cleaned
```

```python
a_Cleaned.columns
```

```python
x=["Income",'Recency','Response','Education','Total_Spent_Amt','Living']
```

```python
sns.pairplot(a_Cleaned[x])
plt.show()
```

```python
data = a_Cleaned.corr(numeric_only=True)
```

```python
plt.figure(figsize=(10,9))
sns.heatmap(data, annot = True, annot_kws= {'size':10},cmap ='coolwarm')
plt.show()
```
